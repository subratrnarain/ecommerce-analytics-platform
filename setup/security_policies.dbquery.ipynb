```sql
-- ############################################################################
-- Filename: setup/security_policies.sql
-- Purpose: Implements Row-Level Security (RLS) for the Silver Layer.
-- ############################################################################

-- 1. Use the correct catalog and schema context
USE CATALOG workspace;
USE SCHEMA dev_silver_layer;

-- 2. Define the Row Filter Function
-- This function acts as a security gatekeeper. In this example, it filters out
-- records where the brand is missing to demonstrate data abstraction.
CREATE OR REPLACE FUNCTION brand_filter(brand_name STRING)
RETURN brand_name IS NOT NULL;

-- 3. Apply the Row Filter to the Silver Table
-- This satisfies the security policy requirement by ensuring all users 
-- interacting with this table are subject to the defined filter logic.
ALTER TABLE events_cleaned 
SET ROW FILTER brand_filter ON (brand);

-- 4. Verify Policy Application
-- Run this to confirm the filter is active in Unity Catalog metadata.
DESCRIBE TABLE EXTENDED events_cleaned;
```
